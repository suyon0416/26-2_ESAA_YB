#### 29. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆)
Hint : use `numpy.add`, `numpy.divide`,`numpy.negative`, `numpy.multiply` and parameter `out` of the funtions

````python
A = np.ones(3, dtype=np.float64) * 1
B = np.ones(3, dtype=np.float64) * 2

# 순서대로 out을 지정하여 메모리 할당 없이 계산
np.add(A, B, out=B)
np.divide(A, 2, out=A)
np.negative(A, out=A)
np.multiply(A, B, out=A)

print(A)
````

#### 30. Extract the integer part of a random array of positive numbers using 2 different methods (★★☆)
Hint : `%`, `//`

````python
Z = np.random.uniform(0, 10, 10)

# 방법 1: % (모듈로 연산자 활용) - 전체 값에서 소수점 아래 버리기
method1 = Z - Z % 1

# 방법 2: // (정수 나눗셈 연산자 활용)
method2 = Z // 1

# 방법 3: np.floor() 함수 사용
method3 = np.floor(Z)

# 방법 4: astype(int) 형변환 사용
method4 = Z.astype(int)

# 방법 5: np.trunc() (소수점 자르기) 사용
method5 = np.trunc(Z)

print("원본:", Z)
print("정수부(방법1):", method1)
print("정수부(방법2):", method2)
````

#### 34. Consider a random 10x2 matrix representing cartesian coordinates, convert them to polar coordinates (★★☆)
Hint 1 : Note that cartesian coodinate **(x, y)** can be represented as  polar coordinates **(distance from origin to (x,y), angle from the x-axis)**

Hint 2 : Use `numpy.sqrt` and `numpy.arctan2`

````python
# 10x2 무작위 직교 좌표계 (x, y) 생성
Z = np.random.random((10, 2))
X, Y = Z[:, 0], Z[:, 1]

# 극좌표계로 변환 (거리 R, 각도 T)
R = np.sqrt(X**2 + Y**2)
T = np.arctan2(Y, X)

print("거리 (R):", R)
print("각도 (T):", T)
````

#### 36. How to find the closest value (to a given scalar v) in a vector Z? (★★☆)
```python
Z = np.arange(100)
v = np.random.uniform(0,100)
```
Hint : Coumpute the distances between the each elements of Z and the scalar v. After that, we can see the index of minimum value using `argmin()`.  

````python
Z = np.arange(100)
v = np.random.uniform(0, 100)

# 1. Z의 각 요소와 v 사이의 절대 거리(abs) 계산 후, 최소 거리가 있는 인덱스 찾기
index = np.abs(Z - v).argmin()

# 2. 해당 인덱스의 실제 값 출력
closest_value = Z[index]

print("기준 값 v:", v)
print("가장 가까운 값:", closest_value)
````

#### 37. What is the equivalent of enumerate for numpy arrays? (★★☆)
Hint : Use `numpy.ndenumerate()`or `numpy.ndindex()`

Example of the output :
```python
Z = np.arange(9).reshape(3,3)
```
```python
# output
(0, 0) 0
(0, 1) 1
(0, 2) 2
(1, 0) 3
(1, 1) 4
(1, 2) 5
(2, 0) 6
(2, 1) 7
(2, 2) 8
```

````python
import numpy as np

Z = np.arange(9).reshape(3, 3)

# 인덱스(index)와 값(x)을 동시에 순회
for index, x in np.ndenumerate(Z):
    print(index, x)
````

#### 38. How to randomly place p elements in a 2D array? (★★☆)
Hint : modify the code below
```python
n = 'size of a 2D array'
p = 'the number of elements that you want to place'
Z = np.zeros((n,n))
np.put(Z, np.random.choice(range(n*n), p, replace = False),'value that you want to place')
print(Z)
```

````python
n = 10  # 2차원 배열의 크기 (10x10)
p = 3   # 무작위로 배치할 요소의 개수

# 1. 10x10 크기의 0으로 채워진 배열 생성
Z = np.zeros((n, n))

# 2. 0부터 (n*n - 1)까지의 인덱스 중 p개를 중복 없이(replace=False) 무작위 선택하여 값(1)을 넣음
np.put(Z, np.random.choice(range(n * n), p, replace=False), 1)

print(Z)
````

#### 39. How to sort an array below by the nth column? (★★☆)
```python
Z = np.random.randint(0,10,(3,3))
```

````python
Z = np.random.randint(0, 10, (3, 3))
print("정렬 전 Z:\n", Z)

n = 1  # 정렬 기준이 될 열의 인덱스 (0: 첫 번째 열, 1: 두 번째 열)

# n번째 열의 값을 기준으로 정렬된 행 인덱스를 구해서 배열 재배열
Z_sorted = Z[Z[:, n].argsort()]

print(f"\n{n}번째 열 기준 정렬 후 Z:\n", Z_sorted)
````
