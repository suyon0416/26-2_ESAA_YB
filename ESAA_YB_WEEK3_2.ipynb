# 12-6 datetime 객체의 메서드

```python
banks=pd.read_csv('/content/drive/MyDrive/ESAA data/banklist.csv',parse_dates=["Closing Date",'Updated Date'])
# parse_dates: 선택한 칼럼들을 datetime 객체로 변환
print(banks.info())
```

```python
banks=banks.assign(
    closing_quarter=banks['Closing Date'].dt.quarter,
    closing_year=banks['Closing Date'].dt.year
)
```

```python
closing_year=banks.groupby(['closing_year']).size()
#closing_year 칼럼에서 같은 값끼리 그룹으로 묶어서 개수 반환
print(closing_year)
```

```python
closing_year_q=(
    banks
    .groupby(['closing_year', 'closing_quarter'])
    .size()
)
print(closing_year_q)
```

```python
ebola=pd.read_csv('/content/drive/MyDrive/ESAA data/country_timeseries.csv', parse_dates=["Date"], index_col="Date")
print(ebola.iloc[:,:4])
```

```python
new_idx=pd.date_range(ebola.index.min(), ebola.index.max())
print(new_idx)
```

```python
new_idx=reversed(new_idx)
print(new_idx)
```

```python
ebola=ebola.reindex(new_idx)
```

```python
print(ebola.iloc[:,:4])
```

```python
last_valid=ebola.apply(pd.Series.last_valid_index)
print(last_valid)
```

```python
earliest_date=ebola.index.min()
print(earliest_date)
```

```python
shift_values=last_valid-earliest_date
print(shift_values)
```

```python
ebola_dict={}

for idx,col in enumerate(ebola):
  d=shift_values[idx].days
  shifted = ebola[col].shift(d)
  ebola_dict[col]=shifted

print(ebola_dict)
```

```python
ebola_shift=pd.DataFrame(ebola_dict)
```

```python
print(ebola_shift.tail())
```

```python
ebola_shift.index=ebola_shift['Day']
ebola_shift=ebola_shift.drop(['Day'], axis="columns")
print(ebola_shift.tail())
```
