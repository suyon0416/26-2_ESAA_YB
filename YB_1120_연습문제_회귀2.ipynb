# **1. 규제 선형 회귀 모델/ 회귀 트리**
- 릿지 회귀모델 및 회귀 트리를 이용하여 전복의 나이를 예측해 보세요.

- 전복의 ‘성별’, ‘키’, ‘지름’, ‘높이’, ‘전체무게’, ‘몸통무게’, ‘내장무게’, ‘껍질무게’를 이용해 ‘껍질의 고리수’를 예측한 뒤, **예측된 ‘껍질의 고리수’에 1.5를 더하면 전복의 나이**가 됩니다.

```python
# 기본 모듈 불러오기
import numpy as np
import pandas as pd

import warnings
warnings.filterwarnings(action='ignore')
```

### 1) 데이터 불러오기 및 데이터 선택

```python
# 데이터 로드
data = pd.read_csv("/content/abalone.csv")

# 성별 M은 Male, F는 Female, I는 Infant 이므로 따로 열 만들기
for label in "MFI":
    data[label] = data["Sex"] == label
data.drop('Sex', axis=1, inplace=True)

# X,y 데이터 선택
y = data.Rings.values
data.drop('Rings', axis=1, inplace=True)

X = data.values.astype(float)
```

## 2) train/test 분리
: 테스트 셋과 데이터 셋을 7:3 비율로 분리하세요.

```python
# 필요한 모듈 불러오기
from sklearn.model_selection import train_test_split
```

```python
# train과 test set 분리 (train:test = 7:3 비율로)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print(X_train.shape, X_test.shape)
```

## 3) 릿지 모델을 이용한 모델링
: 릿지 모델을 이용하여 전복의 나이를 예측해 보고 결과를 평가해 보세요.

 3-1) 릿지모델 생성, 교차검증, 평가에 필요한 모듈을 불러오세요.

```python
#필요한 모듈 불러오기
from sklearn.linear_model import Ridge
from sklearn.model_selection import cross_val_score
from sklearn.metrics import mean_squared_error
```

3-2) 릿지 모델을 생성하고 훈련하여 MSE와 RMSE를 구하세요. 릿지 모델 생성 시  alpha는 0.3으로 설정합니다. (교차 검증 하지 않습니다.)

```python
#릿지모델 생성 및 훈련
ridge = Ridge(alpha=0.3)
ridge.fit(X_train, y_train)

#예측값 구하기
ring_pred = ridge.predict(X_test)


#구한 예측값을 활용해 최종적으로 전복의 나이 예측
age_pred = ring_pred + 1.5
age_true = y_test + 1.5

#MSE와 RMSE구하기
mse = mean_squared_error(age_true, age_pred)
rmse = np.sqrt(mse)
print('MSE : {:.3f}, RMSE : {:.3f}'.format(mse, rmse))
```

3-3) 릿지 모델에서 회귀 계수 값을 구해 알 수 있는 점을 작성하시오. (각 컬럼의 정보는 data.colums 참고)

```python
data.columns
```

```python
coef = pd.Series(ridge.coef_, index=data.columns).sort_values(ascending=False)
print(coef)
```

1. 껍질무게(9.36), 높이(8.89), 전체무게(8.63), 지름(7.95)의 계수가 큰 양수다. 껍질이 무겁고 크기(높이·지름)와 전체 무게가 클수록 고리수가 많아진다. 따라서 이 변수들은 전복의 나이를 높게 예측하는 주요 변수다. 반면 몸통무게(-20.04)와 내장무게(-8.64)는 큰 음수다. 전체무게가 같다면 살과 내장의 비중이 클수록 어린 전복으로 예측한다는 뜻이다. 무게 변수들끼리 상관관계가 강해서 계수가 서로 상쇄되는 다중공선성의 영향도 함께 반영된 결과로 볼 수 있다.

2. 성별 계수는 M(0.31), F(0.22)가 양수, I(-0.53)가 음수다. 다른 조건이 같을 때 유아(Infant) 개체는 성체(수컷·암컷)보다 나이가 적게 예측된다. 수컷과 암컷 사이의 차이는 크지 않다. 또 성별 계수의 절댓값은 무게·크기 변수보다 훨씬 작다. 그래서 전복의 나이 예측에는 성별보다 무게와 크기 같은 신체 측정값의 영향이 훨씬 크다.

3-3) 교차검증을 이용하여 0, 0.1, 1, 10, 100 중  모델 성능이 가장 좋게 나오는 (rmse가 최소가 되는) alpha를 구하세요.

```python
# 주어진 alpha
alphas=[0, 0.1, 1, 10, 100]
```

```python
# alphas list 값을 반복하면서 alpha에 따른 평균 rmse를 구함
for alpha in alphas:
    ridge = Ridge(alpha=alpha)

    # cross_val_score를 이용해 5 폴드의 평균 RMSE를 계산
    neg_mse_scores = cross_val_score(ridge, X_train, y_train,
                                     scoring='neg_mean_squared_error', cv=5)
    avg_rmse = np.mean(np.sqrt(-1 * neg_mse_scores))
    print('alpha {} 일 때 5 folds 의 평균 RMSE : {:.3f}'.format(alpha, avg_rmse))
```

## 4) XGBRegressor을 이용한 모델링
: XGBRegressor 모델을 이용하여 전복의 나이를 예측해 보고 결과를 평가해 보세요.

4-1) XGBRegressor 훈련, 평가에 필요한 모듈을 불러오세요.

```python
# 필요한 모듈 임포트
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error
```

4-2) XGBregressor 모델을 생성하고 훈련하여 MSE와 RMSE를 구하세요.

```python
# 모델 생성 및 훈련
xgb = XGBRegressor(n_estimators=500, learning_rate=0.05, max_depth=4, random_state=42)
xgb.fit(X_train, y_train)

# 예측값 구하기 (껍질의 고리수)
ring_pred_xgb = xgb.predict(X_test)

# 구한 예측값을 활용해 최종적으로 전복의 나이 예측
age_pred_xgb = ring_pred_xgb + 1.5
age_true = y_test + 1.5

# MSE와 RMSE 구하기
mse_xgb = mean_squared_error(age_true, age_pred_xgb)
rmse_xgb = np.sqrt(mse_xgb)
print('MSE : {:.3f}, RMSE : {:.3f}'.format(mse_xgb, rmse_xgb))
```

4-3) 생성한 모델의 피처 중요도를 시각화해보세요.

```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

ftr_importances = pd.Series(xgb.feature_importances_, index=data.columns).sort_values(ascending=False)

plt.figure(figsize=(8, 6))
plt.title('Feature Importances (XGBRegressor)')
sns.barplot(x=ftr_importances.values, y=ftr_importances.index)
plt.show()
```

# **2. 로지스틱 회귀분석**
load_wine() 함수를 이용하여 와인 데이터를 다항분류하는 로지스틱 회귀모델을 생성하세요.

```python
# 필요한 모듈 불러오기
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
```

## 1) 데이터 불러오기 및 변수 선택

```python
wine = load_wine()

wine_x = wine.data # x변수
wine_y = wine.target # y변수
```

## 2) train/test 셋 분리

```python
x_train, x_test, y_train, y_test = train_test_split(wine_x, wine_y, test_size=0.3, random_state=123)

x_train.shape
y_train.shape
```

## 3) 로지스틱 회귀모델을 이용한 모델링

3-1) 로지스틱 회귀모델을 생성하여 훈련하세요.

```python
model = LogisticRegression(max_iter=5000)
model.fit(x_train, y_train)
```

3-2) 회귀 모델의 기울기와 절편을 구하세요.

```python
#기울기
model.coef_
```

```python
#절편
model.intercept_
```

3-3) 모델의 정확도를 구하세요.

```python
#모델의 예측값 구하기

y_pred = model.predict(x_test)
y_pred
#모델의 정확도 구하기

accuracy = metrics.accuracy_score(y_test, y_pred)
print('정확도 : {:.4f}'.format(accuracy))
```
